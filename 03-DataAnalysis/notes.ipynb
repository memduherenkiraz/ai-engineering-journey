## Python Listeleri ve NumPy Performans Karşılaştırması

- Daha iyi anlamak için bir Vücut Kitle İndeksi için hem python listleri hemde numpy kullanalım.
- 100 milyon elemanlı, hepsi aynı değere sahip height ve weight listeleri olsun.
- Python listelerinin ve NumPy dizilerinin aynı işlemdeki performanlarını karşılaştıralım.

```python
import numpy as np
import time
import sys

size = 10 ** 8

# Python listelerinin performansını görmek için bunları kullanacağız.
height_list = [1.82] * size
weight_list = [83.4] * size

# Numpy dizilerinin performansını görmek için aynı verileri numpy dizisine çevirdik.
height_array = np.array(height_list, dtype=np.float32)
weight_array = np.array(weight_list, dtype=np.float32)

# Python List
# time modülü ile başlama ve işlemin bitiş süreleri tutuluyor.
# bmi_list, tüm 100 milyon eleman için VKİ hesaplamasından gelen değerleri tutuyor.

start = time.time()
bmi_list = [w / (h ** 2) for h, w in zip(height_list, weight_list)]
end = time.time()

list_time = end - start   # Geçen süreyi tutar.
list_memory = sys.getsizeof(bmi_list)   # Python listesinin bellekte kapladığı alanı tutar.

print(f"Python Listesi: {list_time:.5f} saniye, Bellek: {list_memory / 1024 / 1024:.2f} MB")

# NumPy
# time modülü ile başlama ve işlemin bitiş süreleri tutuluyor.
# bmi_array, tüm 100 milyon eleman için VKİ hesaplamasından gelen değerleri tutuyor.

start2 = time.time()
bmi_array = weight_array / (height_array ** 2)
end2 = time.time()

array_time = end2 -start2   # Geçen süreyi tutar.
array_memory = sys.getsizeof(bmi_array)   # Numpy dizisinin bellekte kapladığı alanı tutar.

print(f"NumPy: {array_time:.5f} saniye, Bellek: {array_memory / 1024 / 1024:.2f} MB")
```
